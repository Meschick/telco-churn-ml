O que vamos fazer nesse notebook?

- Baixar um CSV do Kaggle
- Carregar em DataFrame
- Validar dados (shape, types, NaN)
- Explorar a variável alvo
- Salvar para próxima etapa

Bibliotecas que vamos usar

- pandas (Manipular os dados)
- numpy (Operações matemáticas)
- opendatasets (baixar os dados) 
- os (gerenciar os caminhos)

```python
import pandas as pd
import numpy as np  
import os
from pathlib import Path
import warnings
```

```python
# Definir caminho base do projeto (funciona em qualquer local)
# Se estiver em Colab, o caminho será diferente
try:
    # Em execução local (VSCode, Jupyter local)
    notebook_dir = Path.cwd()
    
    # Se estiver na pasta notebooks, voltar para raiz
    if notebook_dir.name == 'notebooks':
        project_root = notebook_dir.parent
    else:
        project_root = notebook_dir
    
    print(f"Diretório do projeto: {project_root}")
except:
    # Fallback para Google Colab
    project_root = Path.cwd()
    print(f"Usando diretório: {project_root}")

# Definir caminhos
data_raw_dir = project_root / "data" / "raw"
data_processed_dir = project_root / "data" / "processed"
file_name = "WA_Fn-UseC_-Telco-Customer-Churn.csv"
path_file = data_raw_dir / file_name

print(f"\n{'='*60}")
print("CARREGANDO DATASET")
print(f"{'='*60}")
print(f"Procurando arquivo em: {path_file}")
print(f"Arquivo existe? {path_file.exists()}")

if path_file.exists():
    df = pd.read_csv(path_file, encoding='utf-8')
    print(f"✓ Dataset carregado com sucesso!")
    print(f"\nDimensões: {df.shape[0]} linhas, {df.shape[1]} colunas")
    print(f"\nPrimeiras 5 linhas:")
    print(df.head())
else:
    print(f"\n❌ Arquivo não encontrado!")
    print(f"Certifique-se que o arquivo está em: {path_file}")
    print(f"\nListar arquivos em {data_raw_dir}:")
    if data_raw_dir.exists():
        for file in data_raw_dir.iterdir():
            print(f"  - {file.name}")
    else:
        print(f"  Pasta não existe: {data_raw_dir}")
```

```python
print("\n" + "="*60)
print("1. DIMENSÕES DO DATASET")
print("="*60)
print(f"Linhas: {df.shape[0]}")
print(f"Colunas: {df.shape[1]}")
print(f"\nTotal de células: {df.shape[0] * df.shape[1]:,}")
```

```python
print("\n" + "="*60)
print("2. TIPOS DE DADOS")
print("="*60)
print(df.dtypes)

# Resumo de tipos
print("\nResumo de tipos:")
print(f"  object (string/categórico): {(df.dtypes == 'object').sum()}")
print(f"  int64 (inteiro): {(df.dtypes == 'int64').sum()}")
print(f"  float64 (decimal): {(df.dtypes == 'float64').sum()}")
```

```python
print("\n" + "="*60)
print("3. VALORES AUSENTES (NaN)")
print("="*60)
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100

if missing.sum() > 0:
    print("Colunas com valores ausentes:")
    for col in df.columns:
        if missing[col] > 0:
            print(f"  {col}: {missing[col]} ({missing_pct[col]:.2f}%)")
else:
    print("✓ Nenhum valor ausente encontrado em nenhuma coluna!")
```

```python
print("\n" + "="*60)
print("4. DUPLICADAS")
print("="*60)
duplicates = df.duplicated().sum()
print(f"Total de linhas duplicadas: {duplicates}")

if 'customerID' in df.columns:
    dup_customer_id = df['customerID'].duplicated().sum()
    print(f"customerID duplicados: {dup_customer_id}")
    if dup_customer_id == 0:
        print("✓ Todos os customerID são únicos")
else:
    print("⚠️ Coluna 'customerID' não encontrada")
```

```python
print("\n" + "="*60)
print("5. RESUMO GERAL DO DATASET")
print("="*60)
df.info()
```

```python
print("\n" + "="*60)
print("6. VALIDAÇÃO DE TOTAL CHARGES")
print("="*60)

if 'TotalCharges' in df.columns:
    print(f"Tipo: {df['TotalCharges'].dtype}")
    
    # Tentar converter para numérico
    total_charges_numeric = pd.to_numeric(df['TotalCharges'], errors='coerce')
    invalid_count = total_charges_numeric.isnull().sum() - df['TotalCharges'].isnull().sum()
    
    if invalid_count > 0:
        print(f"❌ Encontrados {invalid_count} valores não-numéricos em TotalCharges")
        print(f"Exemplos de valores inválidos:")
        invalid_mask = pd.to_numeric(df['TotalCharges'], errors='coerce').isnull() & df['TotalCharges'].notna()
        invalid_values = df[invalid_mask]['TotalCharges'].unique()
        for val in invalid_values[:5]:
            print(f"  '{val}'")
    else:
        print("✓ TotalCharges é totalmente numérico (ou vazio)")
else:
    print("⚠️ Coluna 'TotalCharges' não encontrada")
```

```python
print("\n" + "="*60)
print("7. VARIÁVEL ALVO: CHURN")
print("="*60)

if 'Churn' in df.columns:
    print(f"Tipo: {df['Churn'].dtype}")
    print(f"Valores únicos: {df['Churn'].unique()}")
    
    print("\nDistribuição (contagem):")
    churn_counts = df['Churn'].value_counts()
    for category, count in churn_counts.items():
        print(f"  {category}: {count}")
    
    print("\nDistribuição (percentual):")
    churn_pct = df['Churn'].value_counts(normalize=True) * 100
    for category, pct in churn_pct.items():
        print(f"  {category}: {pct:.2f}%")
    
    print("\n⚠️ OBSERVAÇÃO IMPORTANTE:")
    print("   Dados desbalanceados (importante para escolher métrica e estratégia de modelagem)")
else:
    print("❌ Coluna 'Churn' não encontrada!")
```

```python
print("\n" + "="*60)
print("8. SALVANDO DADOS VALIDADOS")
print("="*60)

# Garantir que o diretório existe
data_processed_dir.mkdir(parents=True, exist_ok=True)

output_path = data_processed_dir / "01_raw_validated.csv"
df.to_csv(output_path, index=False)

print(f"✓ Dados salvos com sucesso!")
print(f"\nLocal: {output_path}")
print(f"\nResumo do arquivo salvo:")
print(f"  Linhas: {len(df)}")
print(f"  Colunas: {len(df.columns)}")
print(f"  Tamanho: {output_path.stat().st_size / (1024*1024):.2f} MB")
print(f"\nProxímo passo: Etapa 2 - EDA (02_eda.ipynb)")
```

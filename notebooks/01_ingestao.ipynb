```python
import pandas as pd
import numpy as np  
import os
from pathlib import Path
import warnings
```

```python
# %%
# ============================================================
# DEFINIÇÃO DOS CAMINHOS DO PROJETO
# ============================================================

try:
    notebook_dir = Path.cwd()

    if notebook_dir.name == 'notebooks':
        project_root = notebook_dir.parent
    else:
        project_root = notebook_dir

except Exception:
    project_root = Path.cwd()

data_raw_dir = project_root / "data" / "raw"
data_processed_dir = project_root / "data" / "processed"

data_raw_dir.mkdir(parents=True, exist_ok=True)
data_processed_dir.mkdir(parents=True, exist_ok=True)

file_name = "WA_Fn-UseC_-Telco-Customer-Churn.csv"
path_file = data_raw_dir / file_name

print("=" * 60)
print("CONFIGURAÇÃO DO PROJETO")
print("=" * 60)

print(f"Diretório do projeto: {project_root}")
print(f"Diretório dos dados brutos: {data_raw_dir}")
print(f"Diretório dos dados processados: {data_processed_dir}")
print(f"Arquivo esperado: {path_file}")
```

```python
# %%
# ============================================================
# DOWNLOAD DO DATASET
# ============================================================

import opendatasets as od

kaggle_url = "https://www.kaggle.com/datasets/blastchar/telco-customer-churn"

if not path_file.exists():

    print("Dataset não encontrado localmente.")
    print("Iniciando download do Kaggle...")

    od.download(
        kaggle_url,
        data_dir=str(data_raw_dir)
    )

    print("\nDownload concluído.")

else:

    print("Dataset já encontrado localmente.")
    print("Download não será realizado novamente.")
```

```python
# %%
# ============================================================
# LOCALIZAÇÃO DO DATASET
# ============================================================

possible_files = list(data_raw_dir.rglob(file_name))

if len(possible_files) > 0:

    path_file = possible_files[0]

    print("✓ Dataset encontrado!")
    print(f"\nLocal:")
    print(path_file)

else:

    raise FileNotFoundError(
        f"Não foi possível encontrar o arquivo {file_name} "
        f"dentro de {data_raw_dir}"
    )
```

```python
# %%
# ============================================================
# CARREGAMENTO DO DATASET
# ============================================================

print("=" * 60)
print("CARREGANDO DATASET")
print("=" * 60)

df = pd.read_csv(
    path_file,
    encoding='utf-8'
)

print("✓ Dataset carregado com sucesso!")

print(f"\nDimensões:")
print(f"  Linhas:   {df.shape[0]:,}")
print(f"  Colunas:  {df.shape[1]}")

print("\nPrimeiras 5 linhas:")
display(df.head())
```

```python
print("\n" + "="*60)
print("1. DIMENSÕES DO DATASET")
print("="*60)
print(f"Linhas: {df.shape[0]}")
print(f"Colunas: {df.shape[1]}")
print(f"\nTotal de células: {df.shape[0] * df.shape[1]:,}")
```

```python
# %%
print("=" * 60)
print("2. TIPOS DE DADOS")
print("=" * 60)

print(df.dtypes)

print("\nResumo dos tipos:")

print(
    f"  Object/Categórico: "
    f"{(df.dtypes == 'object').sum()}"
)

print(
    f"  Int64: "
    f"{(df.dtypes == 'int64').sum()}"
)

print(
    f"  Float64: "
    f"{(df.dtypes == 'float64').sum()}"
)
```

```python
# %%
print("=" * 60)
print("3. VALORES AUSENTES")
print("=" * 60)

missing = pd.DataFrame({
    'missing_count': df.isnull().sum(),
    'missing_percent': (
        df.isnull().mean() * 100
    )
})

missing = missing[
    missing['missing_count'] > 0
].sort_values(
    'missing_count',
    ascending=False
)

if len(missing) == 0:

    print("✓ Nenhum valor ausente identificado.")

else:

    print("Colunas com valores ausentes:")
    display(missing)
```

```python
# %%
print("=" * 60)
print("4. DUPLICATAS")
print("=" * 60)

duplicates = df.duplicated().sum()

print(
    f"Linhas completamente duplicadas: {duplicates}"
)

if 'customerID' in df.columns:

    duplicated_ids = df['customerID'].duplicated().sum()

    print(
        f"customerID duplicados: {duplicated_ids}"
    )

    if duplicated_ids == 0:
        print("✓ Todos os customerID são únicos.")
```

```python
# %%
print("=" * 60)
print("5. VALIDAÇÃO DO IDENTIFICADOR")
print("=" * 60)

if 'customerID' in df.columns:

    print("✓ customerID encontrado.")
    print("Essa coluna será removida antes do treinamento,")
    print("pois representa apenas o identificador do cliente.")

else:

    print("⚠️ customerID não encontrado.")
```

```python
print("\n" + "="*60)
print("5. RESUMO GERAL DO DATASET")
print("="*60)
df.info()
```

```python
# %%
print("=" * 60)
print("6. VALIDAÇÃO DE TOTALCHARGES")
print("=" * 60)

if 'TotalCharges' in df.columns:

    total_charges_numeric = pd.to_numeric(
        df['TotalCharges'],
        errors='coerce'
    )

    invalid_mask = (
        total_charges_numeric.isna()
        & df['TotalCharges'].notna()
    )

    invalid_count = invalid_mask.sum()

    print(
        f"Valores não numéricos: {invalid_count}"
    )

    if invalid_count > 0:

        print("\nExemplos:")

        display(
            df.loc[
                invalid_mask,
                ['customerID', 'tenure', 'TotalCharges']
            ].head(10)
        )

    else:

        print(
            "✓ Não foram encontrados valores "
            "não numéricos explícitos."
        )

else:

    print("❌ TotalCharges não encontrado.")
```

```python
# %%
print("=" * 60)
print("7. VALIDAÇÃO DAS VARIÁVEIS CATEGÓRICAS")
print("=" * 60)

categorical_columns = df.select_dtypes(
    include='object'
).columns

for col in categorical_columns:

    if col == 'customerID':
        continue

    print(f"\n{col}")
    print("-" * 40)
    print(
        df[col]
        .value_counts(dropna=False)
        .head(15)
    )
```

```python
# %%
print("=" * 60)
print("8. VARIÁVEL ALVO — CHURN")
print("=" * 60)

if 'Churn' not in df.columns:

    raise ValueError(
        "A coluna Churn não foi encontrada no dataset."
    )

print(f"Tipo: {df['Churn'].dtype}")

print(
    f"\nValores únicos:"
)
print(df['Churn'].unique())

churn_counts = df['Churn'].value_counts()

churn_percent = (
    df['Churn']
    .value_counts(normalize=True)
    * 100
)

target_summary = pd.DataFrame({
    'Quantidade': churn_counts,
    'Percentual': churn_percent.round(2)
})

display(target_summary)

print(
    "\n⚠️ O conjunto apresenta desbalanceamento "
    "entre as classes."
)
```

```python
# %%
print("=" * 60)
print("9. VALIDAÇÃO FINAL DA INGESTÃO")
print("=" * 60)

required_columns = [
    'customerID',
    'gender',
    'SeniorCitizen',
    'Partner',
    'Dependents',
    'tenure',
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies',
    'Contract',
    'PaperlessBilling',
    'PaymentMethod',
    'MonthlyCharges',
    'TotalCharges',
    'Churn'
]

missing_required = [
    col for col in required_columns
    if col not in df.columns
]

if missing_required:

    raise ValueError(
        f"Colunas obrigatórias ausentes: "
        f"{missing_required}"
    )

print("✓ Todas as colunas esperadas estão presentes.")

print(
    f"✓ Dataset possui {df.shape[0]:,} registros."
)

print(
    f"✓ Dataset possui {df.shape[1]} colunas."
)

print(
    f"✓ Variável alvo encontrada: Churn."
)

print(
    "\n✓ INGESTÃO VALIDADA COM SUCESSO."
)
```

```python
# %%
print("=" * 60)
print("10. SALVANDO DADOS VALIDADOS")
print("=" * 60)

output_path = (
    data_processed_dir /
    "01_raw_validated.csv"
)

df.to_csv(
    output_path,
    index=False
)

print("✓ Dados salvos com sucesso!")

print(f"\nArquivo:")
print(output_path)

print(
    f"\nTamanho: "
    f"{output_path.stat().st_size / (1024 * 1024):.2f} MB"
)

print("\nPróximo passo:")
print("Etapa 2 — Análise Exploratória de Dados (EDA)")
```
